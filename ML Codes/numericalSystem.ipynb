# **Bracketing Method is of 2 types**

1.Bisection

2.False position 

## 1. Bisection Method
| Property | Description |
|-----------|--------------|
| **Type** | Bracketing Method |
| **Formula** | $$c = (a + b) / 2$$ |
| **Update Rule** | If f(c) > 0 → b = c; else a = c |
| **Convergence** | Linear (slow but guaranteed) |

**Equation :$$x^2=2$$**

```python
import math as m

f = lambda x: x**2-2
a=1
b=2
realVal= m.sqrt(2)
tolerence =0.001         #!   how much error can be tolareted

for i in range(1,101):
    c=(a+b)/2
    err = abs(realVal-c)
    print(f"iter {i}: approx. root= {c:.4f} error= {err:.4f}")
    if err <tolerence:
        break
    if f(c)>0:
        b=c
    else :
        a=c
```

## 2.False Position(reguli falsi) Method
| property | Description |
|----------|--------------|
| Type | Bracketing method |
| Formula | $$c = \frac{a f(b) - b f(a)}{f(b) - f(a)}$$ |
| Update Rule |	If ( f(c) > 0 ), then ( b = c ); else ( a = c ) |
| Convergence |	Linear (faster than bisection) |

* Equation :$$x^2=2$$

```python
import math as m
f = lambda x: x**2 - 2
a =1
b =2
realVal = m.sqrt(2)
tolerence = 0.001
for i in range(1,100):
    c= (a*f(b)-b*f(a))/(f(b)-f(a))
    err = abs(realVal-c)
    print(f"iter {i}: approx._root= {c:.6f} error= {err:.6f}")
    if err <tolerence:
        break
    if f(c)>0:
        b=c
    else :
        a=c
```

# **Open method is of 3 types**

1.Secant

2.Fixed point
 
3.Newton Raphson

## **1. Secant Method**

| **Property**    | **Description**                                                           |
| --------------- | ------------------------------------------------------------------------- |
| **Type**        | Open Method                                                               |
| **Formula**     | $$c = \frac{a f(b) - b f(a)}{f(b) - f(a)}$$ |
| **Update Rule** | a=0, b=1 Then in loop set: a = b, b = c      |
| **Convergence** | Superlinear                        |


* Equation
$$e^{-x} - x=0$$

```python
import math as m
f = lambda x:m.exp(-x)-x
a = 0
b = 1
for i in range(1,6):
    c = (a*f(b)-b*f(a))/ (f(b)-f(a))
    print(f"iter {i}: Approx._root: {c:.4f}")
    a=b
    b=c
```

# **2. Fixed point iteration**
|Property |	Description |	
|---------|--------------|	
|Type |	Open Iterative Method |	
|Formula | $$x_1 = g(x_0)$$ |		
|Convergence |	Linear |
|Key Feature | Simple to use; convergence depends on g(x) choice. $$g'(x)<1$$ |	

* Equation
$$e^{-x} - x=0$$

```python
import math as m
import matplotlib.pyplot as plt
root =[]
error =[]
g= lambda x: m.exp(-x)
x0 = 0
for i in range(15):
    x1= g(x0)
    root.append(x1)
    
    err = abs(x1-x0)
    error.append(err)
    x0=x1
plt.plot(root)
plt.plot(error)
plt.show()
```

* Equation
$$x^{2} - sinx =0$$

```python
import math as m
import matplotlib.pyplot as plt
root =[]
g= lambda x: m.sqrt(m.sin(x))
x0 = 1
error =[]
for i in range(11):
    x1= g(x0)
    root.append(round(x1,4))
    err = abs(x1-x0)
    error.append(err)
    x0=x1
print(root)
plt.plot(root,label ='root')
plt.plot(error,label ='error')
plt.legend()
plt.show()
```

# **3.Newton-Raphson Method**

| **Property**    | **Description**                                                      |
| --------------- | -------------------------------------------------------------------- |
| **Type**        | Open Derivative Method                                               |
| **Formula**     | $$xₙ₊₁ = xₙ - \frac{f(xₙ)} {f'(xₙ)}$$                          |
| **Condition**   | Requires $$f'(x)!=0$$                                           |
| **Update Rule** | $$x_1 = x_0 - \frac{f(x_0)}{df(x_0)}$$            |
| **Convergence** | Quadratic (very fast)                                                |
| **Key Feature** | Extremely fast if initial guess is close; sensitive to poor guesses. |

- Equation 
$$e^{-x}-x=0$$

```python
import math

f=lambda x: math.exp(-x)-x
df=lambda x: -math.exp(-x)-1
x0=0
iter=1
err =1
while iter<=5:
    x1=x0-f(x0)/df(x0)
    err=abs(x1-x0)
    print(" Iter: %d  App_root: %0.5f Abs_err:%f"%(iter,x1,err))
    x0=x1
    iter=iter+1
```

# Trapezoidal Method 

```python

import matplotlib.pyplot as plt
import math as m 

f=lambda x: m.exp(x)
a=0
b=1
n=4
h=(b-a)/n

sum = f(a)+f(b)

for i in range(n-1):
    a=a+h
    sum=sum+(2*f(a))
    
sum = sum*h/2
    
print(sum)
```


# Simpsons 1/3 Rule

```python

import matplotlib.pyplot as plt
import math as m 

f=lambda x: m.exp(x)

n=10
a=0
b=1

h=(b-a)/n
sum= f(a)+f(b)

for i in range (1,n):
    a = a+h
    if i%2==0:
        sum=sum+(2*f(a))
    else:
        sum=sum+(4*f(a))
        
sum = sum*(h/3)
print(sum)
```
