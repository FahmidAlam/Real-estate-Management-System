# Pandas mainly has 2 primary Data Structures:
    -Series (1D, lebeled,  homogeneous, size-immutable)
    -DataFrame (2D, lebeled, heterogeneous, size-mutable)

Series :-

```python
import pandas as pd
x = pd.Series(data=['Fahmid',22,5.7,'Male'],index=['Name','Age','Height','Gender'])
print(x)
```

DataFrame:-

```python
import pandas as pd
data ={'Name': ['fahmid','nafis','ishfaq','azraf','jarraf'],
    'Age': [22,23,21,23,24],
    'Gender': 'male'
    }
df = pd.DataFrame({
    'Name': ['fahmid','nafis','ishfaq','azraf','jarraf'],
    'Age': [22,23,21,23,24],
    'Gender': 'male'
})
print(df)
#print(df["Name"])
```

```python
import pandas as pd
import numpy as np
n= np.array([1,2,3])
s = pd.Series([1,2,3])
result = s.to_list()
print(n)
print(type(n))
print('output:',result)
print('output type:',type(result))
print(*s)
print(type(s))
```
