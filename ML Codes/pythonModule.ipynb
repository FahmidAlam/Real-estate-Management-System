# Ways of importing Modules

Four ways to import modules

```python
import math as m
m.sqrt(5)

```

```python
from math import sqrt
sqrt(4)
```

```python
from math import *
sqrt(112)
```

```python
from math import sqrt as s
s(64)
```

```python
import math
help(math.sqrt)
```

```python
li = [1,2,3,4]
li.extend(5)
```
