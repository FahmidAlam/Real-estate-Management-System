# Regex(regular expression):

A special syntax for searching for strings that meets a specified pattern. It's a great tool to filter and sort through text when you want to match patterns rather than a hard coded string or strings.

```python
import re
```

```python
result_search = re.search('pattern',r"string containing th pattern")
print(result_search)
print(result_search.group())          #? only retruns the matching pattern
```

```python
string = r'shadhen tries to help us get motivated towards cp'
new_string = re.sub(r'shadhen',r'swadhen',string)
print(new_string)
```

```python
customer_reviews = ['sam was a great help to me in the store', 
                    'the cashier was very rude to me, I think her name was eleanor', 
                    'amazing work from sadeen!', 
                    'sarah was able to help me find the items i needed quickly', 
                    'lucy is such a great addition to the team', 
                    'great service from sara she found me what i wanted'
                    ]
```

```python
sarahs_review = []
pattern_to_find = r"sarah?"
```

```python
for string in customer_reviews:
    if(re.search(pattern_to_find,string)):
        sarahs_review.append(string)
(sarahs_review)
```

- Finding strings that starts with a specific char

```python

a_reviews = []
pattern_to_find = r"^a"      #? the ^ operator to indicates the start of a string


for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        a_reviews.append(string)
a_reviews
```

- Finding strings that ends with a specific char

```python
y_reviews = []
pattern_to_find = r"y$"          #? the $ operator to indicate the end of a string

for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        y_reviews.append(string)

y_reviews
```

```python
needwant_reviews = []
pattern_to_find = r"(need|want)ed" #! the pipe operator | can be used to mean OR
for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        needwant_reviews.append(string)
needwant_reviews
```

- Removing punctuation

```python
no_punct_reviews = []
pattern_to_find = r"[^\w\s]" 
#! [^ ] means "not", \w means word and \s means whitespace: so find anything that is not a word or a space
for string in customer_reviews:
    no_punct_string = re.sub(pattern_to_find, "", string)
    no_punct_reviews.append(no_punct_string)
print(no_punct_reviews)
```

# Stemming
- removes suffix of words

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
connect_tokens= ['connecting','connected','conectivity','connects','connectted']
for i in connect_tokens:
    print(i,' : ',ps.stem(i))
```

```python
like_tokens=['likes','beTtEred','worse']
for i in like_tokens:
    print(i,' : ',ps.stem(i))
```

# Lemmatization

```python

nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python

lemmatizer = WordNetLemmatizer()
for i in connect_tokens:
    print(i,' : ',lemmatizer.lemmatize(i))
```

```python
for i in like_tokens:
    print(i,' : ',lemmatizer.lemmatize(i))
```

# **n-grams**
- Breaks string then groups words or tokens into sequence based on 'n' sequence similarity

```python
import nltk
import pandas as pd
import matplotlib.pyplot as plt
```

```python
customer_reviews = '''The environment encompasses all living and non-living things occurring naturally on 
Earth or in some region. It includes all the biotic and abiotic factors influencing each other in nature. 
All the living elements like birds, animals, plants, forests, etc. comprise the biotic elements. 
Everything non-living like air, water, rocks, sun, etc. is an example of the abiotic 
component of the environment.'''

tokens =  word_tokenize(customer_reviews)
print(tokens)
```

```python
words = [
    "cat", "sits", "on", "the", "mat",
    "dog", "runs", "fast", "the", "the",
    "bird", "flies", "high", "sun", "shines",
    "bright", "rain", "falls", "slowly",
    "night", "comes", "cat", "cat", "dog",
    "wind", "blows", "tree", "moves", "the",
    "sky", "blue", "sun", "sun", "sun"
]
```

```python
unigrams =(pd.Series(nltk.ngrams(words,1)).value_counts())
print(unigrams[:10])
unigrams[:10].sort_values().plot.barh(color='maroon',figsize=(8, 5))
```

```python
bigrams =(pd.Series(nltk.ngrams(words,2)).value_counts())
print(bigrams[:10])
bigrams[:10].sort_values().plot.barh(color='maroon',figsize=(8, 5))
```

```python
trigrams =(pd.Series(nltk.ngrams(words,3)).value_counts())
print(trigrams[:10])
trigrams[:10].sort_values().plot.barh(color='maroon',figsize=(8, 5))
```
