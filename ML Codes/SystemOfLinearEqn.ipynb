## **1. Gauss–Jacobi Method**

| **Property**    | **Description**                                                                     |
| --------------- | ----------------------------------------------------------------------------------- |
| **Type**        | Iterative Linear System Solver                                                      |
| **Update Rule**     | a,b,c = 0,0,0  then in loop x,y,z equations using a,b,c then a,b,c = x,y,z |
| **Convergence** | Linear                                                                              |

* Equation: $$3x+y+z=5$$ $$2x-5y+z=20$$ $$x+y+z=10$$ 

```python
import matplotlib.pyplot as plt

a, b, c = 0, 0, 0

# store values for plotting
xs, ys, zs = [], [], []
iters = []

for i in range(1, 10):
    x = (5 - b - c) / 3
    y = (2 * a + c - 20) / 5
    z = (10 - a - b) / 7

    a, b, c = x, y, z

    print(f"iter {i}: x={a:.3f} y={b:.3f} z={c:.3f}")

    xs.append(a)
    ys.append(b)
    zs.append(c)
    iters.append(i)

# -------------------------------
# Plotting
# -------------------------------
plt.plot( xs, label="x values")
plt.plot(ys, label="y values")
plt.plot(zs, label="z values")

plt.xlabel("Iterations")
plt.ylabel("Variable Values")
plt.title("Convergence of x, y, z")
plt.legend()
#plt.grid(True)

plt.show()
```

# 2.Gauss Seidal :


| **Property**    | **Description**                                                                                                |
| --------------- | -------------------------------------------------------------------------------------------------------------- |
| **Type**        | Iterative Linear System Solver                                                                                 |
| **Update Rule**     | x,y,z = 0,0,0  Then in loop x,y,z equations |
| **Convergence** | Linear (faster than Jacobi)                                                                                    |

```python
x,y,z = 0, 0, 0  # initial guesses

for i in range(10):
    # compute new values using ONLY old values
    x = (5 - y - z) / 3
    y = (2*x + z - 20) / 5
    z = (10 - x - y) / 7


    print(f"Iter={i+1} x={x:.3} y={y:.3} z={z:.3}")
```

```python
import matplotlib.pyplot as plt

a, b, c = 0, 0, 0  # initial guesses
x_vals, y_vals, z_vals = [], [], []  # store values for plotting

for i in range(10):
    # compute new values using ONLY old values
    x = (5 - b - c) / 3
    y = (2*a + c - 20) / 5
    z = (10 - a - b) / 7
    
    # update all together
    a, b, c = x, y, z
    
    # store for plotting
    x_vals.append(x)
    y_vals.append(y)
    z_vals.append(z)
    
    print(f"Iter={i+1} x={x:.3} y={y:.3} z={z:.3}")

# plotting
plt.figure(figsize=(10, 6))
plt.plot(x_vals, 'o-', label='x')
plt.plot( y_vals, 's-', label='y')
plt.plot( z_vals,'x-',  label='z')
plt.xlabel('Iteration')
plt.ylabel('Value')
plt.title('Convergence of x, y, z')
plt.legend()
plt.grid(True)
plt.show()
```

# **Interpolation Methods**
- Finds approximate value of $f(x)$ for unkown value of x

***Types :-***
1. Lagrange method
2. Linear spline method
3. Direct method
4. Newtonn Divided method

## **1. Lagrange Interpolation**

| **Property**         | **Description**                                                                                    |
| -------------------- | -------------------------------------------------------------------------------------------------- |
| **Type**             | Polynomial Interpolation                                                                           |
| **Formula**          | $P(x) = Σ [ yᵢ * Lᵢ(x) ]$ , i = 0 to n |
| **Key Feature**      | Simple and direct; can oscillate for high-degree polynomials.                                      |

```python
import math as m

fun = lambda p: 2**p

x = [1, 2, 3]
y = [4, 5, 7]
n = len(x)
xx = 2.5
exact = fun(xx)

f = 0
for i in range(n):
    L = 1
    for j in range(n):
        if(i!=j):
            L *= ((xx-x[j])/(x[i]-x[j]))
    f += L*y[i]

print(f'Exact value ={exact}')
print("Interpolated value at x =", xx, "is y =", f)
err=abs(exact-f)/exact*100
print(f'Error={err}%')
```

## **2. Linear Spline Interpolation**

| **Property**    | **Description**                                           |
| --------------- | --------------------------------------------------------- |
| **Type**        | Piecewise Linear Interpolation                            |
| **Formula**     | For interval x[i], x[i+1]:$S = y[i] + (y[i + 1] - y[i]) * (xx - x[i]) / (x[i + 1] - x[i])$ |

```python
x = [1, 2, 3, 4]
y = [4, 5, 7, 8]

n = len(x)
xx = 2.5

for i in range(n - 1):
    if (xx >= x[i]) and (xx <= x[i + 1]):
        S = y[i] + (y[i + 1] - y[i]) * (xx - x[i]) / (x[i + 1] - x[i])
        print("Interpolated value:", S)
```

## **3. Newton’s Divided Difference**

| **Property**           | **Description**                                                              |
| ---------------------- | ---------------------------------------------------------------------------- |
| **Type**               | Polynomial Interpolation                                                     |
| **Formula**            | $F[i][j] = (F[i + 1][j - 1] - F[i][j - 1]) / (x[i + j] - x[i])$|
| **Purpose**            | Build interpolating polynomial recursively                                   |

```python
import math as m
x = [1, 2, 3]
y = [4, 5, 7]
n = len(x)
xx = 1.5

# Initialize divided difference table
F = []
for i in range(n):
    F.append([0]*n)  #F = [[0 for _ in range(n)] for _ in range(n)]
    F[i][0] = y[i]

# Compute divided differences
for j in range(1, n):
    for i in range(n - j):
        F[i][j] = (F[i + 1][j - 1] - F[i][j - 1]) / (x[i + j] - x[i])

# Newton interpolation
p = F[0][0]
product = 1
for i in range(1, n):
    product *= (xx - x[i - 1])
    p += F[0][i] * product
func = lambda x : 2 ** x
trueValue = func(xx)
abs_err = abs(trueValue - p)

# multiply by (x - x[i-1])
# add current term to p
per_err = abs(((trueValue - p) / trueValue) * 100)
print(f"Interpolated value at x = {xx} is {p:0.6f}\nAbs.Error: {abs_err:.5f}\n%Error: {per_err:.5f}")
```
